### Problem 2: 온도변환 프로그램
😄화씨온도와 섭씨온도 간 변환공식은 아래와 같다. 화씨온도 100은 섭씨온도 몇인가? 반대로 섭씨온도 50은 화씨온도 몇인가?
화씨온도를 저장할 변수 f와 섭씨온도를 저장할 변수 c를 만들고
변환공식을 이용하여 코드를 작성하고 결과를 출력해보세요.

변환 공식

<font color='blue'>`화씨(f) → 섭씨(c)`</font>
```python
c = 5.0 / 9.0 * (f - 32.0)
```
<font color='blue'>`섭씨(c) → 화씨(f)`</font>
```python
f = 9.0 / 5.0 * c + 32.0
```
---

Output Sample
```
37.77777777777778
```
Output Sample
```
122.0
```

📜 참고: 코드를 실행하면 화씨온도 100이 섭씨온도 약 37.78이고 섭씨온도 50이 화씨온도 122임을 알 수 있다.
복잡한 연산에서는 괄호의 위치에 특히 유의해야 한다.

````python
#화씨(f) → 섭씨(c)
f = 100
c = (f - 32) * 5 / 9
print(c)
````

````python
#섭씨(c) → 화씨(f)
c = 50
f = c * 9 / 5 + 32
print(f)
````

### (과제) Problem 5: 지불 금액 계산2

😄 우유의 가격은 1850원이다. 계란 한판의 가격은 4000원이다. 모든 제품은 5%를 할인해준다. 우유 2개, 계란 한판을 구입했을 때 지불해야 할 가격을 변수를 활용하여 코드로 작성하시오.

---
Output

```
7315.0
```

📜 참고: 코드를 실행하면 7315.0 이 출력되는데 이 값이 지불해야 할 가격이다. 변수 이름에 공백이 들어가지 않도록 유의한다.

````python
milk_price = 1850
egg_price = 4000

milk_cnt = 2
egg_cnt = 1

discount_rate = 0.05

subtotal = milk_price * milk_cnt + egg_price * egg_cnt
total = subtotal * (1 - discount_rate)

print(total)
````

### (과제) Problem 9: 비밀 금고 해킹 코드 만들기 🏦

😄 금고에는 사전에 정해진 큰 수 하나가 저장되어 있어요.

이 숫자에서 규칙대로 숫자들을 뽑아 “해킹 코드”를 출력해보세요.

아래 값은 코드 안에 미리 정의되어 있어야 합니다.

```
SAFE = 20260112
```

다음 규칙으로 4개의 숫자를 만들어 한 줄로 출력하세요.

- A : SAFE의 마지막 2자리
- B : SAFE의 끝에서 3번째 자리(백의 자리)
- C : SAFE를 1000으로 나눈 몫의 마지막 1자리
- D : SAFE를 10000으로 나눈 몫의 마지막 2자리

출력 형식은 아래와 같습니다:
```
A-B-C-D
```
---


Output Sample
```
12-1-0-26
```

````python
SAFE = 20260112

A = SAFE % 100                 # 마지막 2자리
B = (SAFE // 100) % 10         # 백의 자리
C = (SAFE // 1000) % 10        # 천 단위로 자른 뒤 마지막 1자리
D = (SAFE // 10000) % 100      # 만 단위로 자른 뒤 마지막 2자리

print(f"{A}-{B}-{C}-{D}")
````
