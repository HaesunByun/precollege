### Problem 2: 시험 성적 관리하기

열 명의 학생이 국어, 영어, 수학 총 세 과목의 시험을 보았다.  
학생의 이름과 과목별 시험 결과는 별개의 리스트에 순서대로 저장되어 있다.  
딕셔너리를 이용해서 학생들의 성적을 정리해 보자.

---



````python
# 먼저 이 블록을 실행하자.
name = ['민준', '하윤', '준서', '지윤', '서아', '은우', '소율', '유찬', '가은', '준영']
scores_kor = [92, 85, 60, 54, 64, 63, 67, 90, 99, 90]
scores_eng = [97, 90, 95, 71, 77, 59, 50, 67, 78, 92]
scores_math = [62, 53, 57, 54, 98, 81, 99, 97, 69, 96]
````

<font color='blue'>`Task 2-1`</font>

이제, 두 딕셔너리(scores1, scores2)를 각기 다른 방법으로 구성할 것이다.
<br><br>
scores1은 '이름', '국어', '영어', '수학' 총 네 개의 Key와 각각에 해당하는 리스트 Value로 구성된다.
<br><br>
scores2는 학생들의 이름을 Key로, 각 과목의 성적이 저장된 딕셔너리를 Value로 가진다. 즉, 딕셔너리 안에 딕셔너리가 들어 있다.  
scores2의 Value에 해당하는 작은 딕셔너리들은, '국어', '영어', '수학' 총 세 개의 Key와 각각의 점수 Value로 구성된다.
<br><br>
조건에 맞는 딕셔너리 scores1, scores2를 만들고, 각 딕셔너리를 출력하자.
<br><br>
Output Sample
```
{'이름': ['민준', '하윤', '준서', '지윤', '서아', '은우', '소율', '유찬', '가은', '준영'], '국어': [92, 85, 60, 54, 64, 63, 67, 90, 99, 90], '영어': [97, 90, 95, 71, 77, 59, 50, 67, 78, 92], '수학': [62, 53, 57, 54, 98, 81, 99, 97, 69, 96]}
{'민준': {'국어': 92, '영어': 97, '수학': 62}, '하윤': {'국어': 85, '영어': 90, '수학': 53}, '준서': {'국어': 60, '영어': 95, '수학': 57}, '지윤': {'국어': 54, '영어': 71, '수학': 54}, '서아': {'국어': 64, '영어': 77, '수학': 98}, '은우': {'국어': 63, '영어': 59, '수학': 81}, '소율': {'국어': 67, '영어': 50, '수학': 99}, '유찬': {'국어': 90, '영어': 67, '수학': 97}, '가은': {'국어': 99, '영어': 78, '수학': 69}, '준영': {'국어': 90, '영어': 92, '수학': 96}}
```

````python
scores1 = {}
scores1['이름'] = name
scores1['국어'] = scores_kor
scores1['영어'] = scores_eng
scores1['수학'] = scores_math


scores2 = {}
for i in range(len(name)):
  scores2[name[i]] = { '국어' : scores_kor[i], '영어' : scores_eng[i], '수학' : scores_math[i] }


print(scores1)
print(scores2)
````

````python
###########################################################################
# (선택) 딕셔너리를 보기 좋게 출력해 주는 함수를 import 해서 사용해 보자.
from pprint import pprint

print('scores1\n')
pprint(scores1)

print('\n\nscores2\n')
pprint(scores2)
###########################################################################
````

<font color='blue'>`Task 2-2`</font>

두 딕셔너리를 각각 이용해서 하윤이의 세 과목 성적을 확인해 보자.
<br><br>
Output sample
```

```

````python
ind_hy = scores1['이름'].index('하윤')
print(f"scores1\n국어: {scores1['국어'][ind_hy]}, 영어: {scores1['영어'][ind_hy]}, 수학: {scores1['수학'][ind_hy]}")

print()
print(f"scores2\n국어: {scores2['하윤']['국어']}, 영어: {scores2['하윤']['영어']}, 수학: {scores2['하윤']['수학']}")
````

<font color='blue'>`Task 2-3`</font>

두 딕셔너리를 각각 이용해서 민준이의 세 과목 평균을 구해 보자.  
평균은 소수점 셋째 자리에서 반올림하자.
<br><br>
Output Sample
```
scores1: 83.67
scores2: 83.67
```

````python
ind_mj = scores1['이름'].index('민준')
avg1 = (scores1['국어'][ind_mj] + scores1['영어'][ind_mj] + scores1['수학'][ind_mj]) / 3
print(f"scores1: {round(avg1,2)}")

avg2 = (scores2['민준']['국어'] + scores2['민준']['영어'] + scores2['민준']['수학']) / 3
print(f"scores2: {round(avg1,2)}")
````

<font color='blue'>`Task 2-4`</font>

두 딕셔너리 각각에 학생들의 세 과목 평균 점수를 "평균"이라는 키로 저장하고, 딕셔너리를 출력해 보자.  
평균은 소수점 셋째 자리에서 반올림하자.

````python
scores1['평균'] = []
for i in range(len(scores1['이름'])):
  avg = (scores1['국어'][i] + scores1['영어'][i] + scores1['수학'][i])/3
  scores1['평균'].append(round(avg,2))

for name in scores2.keys():
  avg = (scores2[name]['국어'] + scores2[name]['영어'] + scores2[name]['수학'])/3
  scores2[name]['평균'] = round(avg,2)

print(scores1)
print(scores2)
````

### Problem 3: 투표 결과 정리하기

회장선거 결과를 딕셔너리로 정리해 보자.  
총 선거인수는 200명이고, 투표 결과는 inp_vote라는 변수명에 다음과 같은 문자열 형식으로 저장되어 있다.  
아래의 예시는 총 선거인수가 10명인 경우이다.

---

````python
# 예시: 총 선거인수가 10명인 경우: 먼저 이 블록을 실행하자.
inp_vote = "지우 예준 서준 지우 서준 도현 도현 지우 도현 도현"
````

````python
#################################################################################################
# 이 블록의 코드는 수정하지 말 것.
# 선거인수가 num_vote명일 때의 투표 결과를 랜덤하게 생성하고 inp_vote 변수에 저장하는 코드이다.
# 일단 선거인수가 10명인 상황 (위 블록)에 대해 코드를 완성하자.
# 그 이후 이 블록을 실행시켜 선거인수가 200명인 경우에 대해서도 확인해 보자.

from random import random

num_vote = 200

inp_vote = ""
people = "서준 지우 예준 하윤 도현".split()
for i in range(num_vote): inp_vote = inp_vote + " " + people[int(random()*5)]

#################################################################################################
````

<font color='blue'>`Task 3`</font>



먼저 빈 딕셔너리를 하나 만들자. 이 딕셔너리에는 후보들의 이름이 키(Key)로, 득표수가 값(Value)으로 들어갈 것이다.  
후보는 서준, 지우, 예준, 하윤, 도현 총 다섯 명이다.
<br><br>
다음으로, 투표 결과 문자열을 리스트로 변환하자.  
문자열의 split() 메서드를 사용하면 각 원소가 후보의 이름인 리스트를 생성할 수 있다.
<br><br>
이제 리스트의 원소(이름)를 하나씩 확인한다.  
이름 키에 대응하는 값에 1을 더하는 과정을 리스트의 모든 원소에 대해 반복하자.  
이후 딕셔너리를 출력하고, 총 투표수를 구해 보자.



````python
votes = inp_vote.split()
resvote = { '서준': 0, '지우': 0, '예준': 0, '하윤': 0, '도현': 0 }


for name in votes:
  resvote[name] += 1

print(resvote)


sum_votes = 0
for i in resvote.keys():
  sum_votes += resvote[i]
print(f"총 투표수: {sum_votes}명")
````

### Problem 4: 술게임 구현하기

술게임 중 하나인 "더 게임 오브 데스"를 딕셔너리로 구현해 보자.
<br><br>
------------------------게임 설명-------------------------

<font color='blue'>`Task 4-1`</font>

먼저 게임에 참여하는 사람들의 이름이 names 리스트로 제시된다.
```
names = ['서준', '지우', '예준', '하윤', '도현']
```
각 인원의 이름을 Key로, 그 사람이 가리키는 사람의 이름을 Value로 하는 딕셔너리 god(game of death)가 주어진다.

```
god = {         
  '서준': '지우',
  '지우': '예준',
  '예준': '하윤',
  '하윤': '도현',
  '도현': '서준'
  }
```
<br>시작하는 사람은 서준이로 정한다.
"시작하는 사람: 서준" 이라는 문자열을 출력한다.
<br><br>
"1보다 큰 정수 입력: " 이라고 출력하고, 정수를 입력받는다.
<br><br>
시작하는 사람의 이름 Key에 대응되는 Value를 확인한다. 그 사람의 이름이 다음 Key가 된다.
<br>
입력받은 숫자만큼 이 과정을 반복한다.
<br><br>
마지막으로 지목된 사람을 확인한다.
"걸린 사람: XX" 라고 출력한다.
<br><br>
마지막으로 딕셔너리를 출력해 누가 누구를 가리키고 있는지 확인한다.



````python
names = ['서준', '지우', '예준', '하윤', '도현']
god = { '서준': '지우', '지우': '예준', '예준': '하윤', '하윤': '도현', '도현': '서준' }

print(f"시작하는 사람: 서준")
name = '서준'

num = int(input("1보다 큰 정수 입력: "))

for i in range(num): name = god[name]

print(f"걸린 사람: {name}")

print(god)
````

<font color='blue'>`Task 4-2`</font>

이번에는 사람들의 이름, 인원 수, 시작하는 사람이 모두 입력값으로 들어오는 경우에 대해 코드를 작성해 보자. <br><br>

먼저 "참여 인원:" 이라고 출력하고, 게임에 참여하는 사람들의 이름을 입력받자.<br>
참여자의 이름은 공백으로 구분되어 하나의 문자열로 입력된다.  인원 수는 정해져 있지 않다.<br>
입력된 문자열의 이름들을 원소로 가지는 리스트 names를 생성하자. 입력 예시는 다음과 같다.
```
서준 지우 예준 하윤 도현
```

<br>이제 참여자들의 이름을 Key로, 그 사람이 가리키는 사람의 이름을 Value로 하는 딕셔너리 god를 만들자.<br>
Value는 names 리스트 안의 원소 중 하나를 랜덤으로 선택하자. <br>
(선택 사항: 자기 자신을 가리키는 경우가 없도록 만들어 보자. 코드를 어떻게 수정해야 할까?)
<br><br>
"시작하는 사람: " 이라고 출력하고, 이름을 입력받는다.
<br><br>
"1보다 큰 정수 입력: " 이라고 출력하고, 정수를 입력받는다.
<br><br>
마지막으로 지목된 사람을 확인한다. "걸린 사람: XX" 라고 출력한다.
<br><br>
마지막으로 딕셔너리를 출력해 누가 누구를 가리키고 있는지 확인한다.


````python
from random import random

names = input("참여 인원: ").split()
name = input("시작하는 사람: ")
num = int(input("1보다 큰 정수: "))

god = {}
for i in range(len(names)):
  god[names[i]] = names[(i + 1 + int(random()*(len(names)-1)))%len(names)]
  god[names[i]] = names[(i + 1)%len(names)]

for i in range(num):
    name = god[name]


print(f"걸린 사람: {name}")
print(god)
````

투표 결과는 inp_vote라는 변수명에 Problem 3의 경우와 같은 문자열 형식으로 입력받는다.
<br><br>
리스트의 원소를 하나씩 확인할 때 이름이 딕셔너리의 키로 저장되어 있다면 대응하는 값에 1을 더하고,<br>
저장되어 있지 않다면 새로운 키와 값을 추가하자. <br>
리스트의 모든 원소에 대해 반복한 후 딕셔너리를 출력하고, 총 투표수를 구해 보자.

입력 예시는 다음과 같다.
```
예준 예준 도현 예준 예준 하윤 도현 서준 지우
```

````python
inp_vote = input()

resvote = {}

votes = inp_vote.split()

for name in votes:
    if name in resvote:
        resvote[name] += 1
    else:
        resvote[name] = 1
print(resvote)


sum_votes = 0
for i in resvote.keys():
  sum_votes += resvote[i]
print(f"총 투표수: {sum_votes}명")
````

### Homework 3: 꼬리 잡기 시뮬레이션

꼬리 잡기 게임을 구현해 보자.
<br>
먼저 게임에 참여하는 사람들의 이름이 공백으로 구분된 문자열로 제시된다.  
이후, 참여자들의 이름을 원소로 가지는 names 리스트를 만들자.

```
서준 지우 예준 하윤 도현
```

게임 시작 전 각자가 잡아야 하는 사람이 랜덤으로 정해진다. <br> 이 관계는 게임 시작 전에 출력된다. 다음은 출력 예시이다.
```
지우 -> 예준 -> 도현 -> 하윤 -> 서준 -> 지우
```

단, 모든 인원에게는 단 한 명의 목표만 배정되어야 하며, 쫓고 쫓기는 관계는 하나의 고리로 연결되어야 한다. <br>
(힌트: random 라이브러리의 shuffle 함수를 사용하면 리스트의 순서를 섞을 수 있다.)

````python
# shuffle 함수는 리스트의 순서를 섞는다.
from random import shuffle

name = [1, 2, 3, 4, 5]
shuffle(name)

print(name)
````

게임이 시작되면, 매 라운드마다 랜덤으로 한 명이 선택된다. 선택된 사람은 자신의 목표를 잡은 것으로 간주한다. <br>
자신의 목표를 잡은 사람의 다음 타겟은, 방금 잡은 사람의 목표로 다시 지정된다.<br><br>
예를 들어, 위의 상황에서 하윤이가 선택되었다면 하윤이가 서준이를 잡은 상황으로 간주한다. <br> 따라서 서준이는 게임에서 제외되고, 이후 하윤이의 목표는 지우가 된다.
<br><br>
매 라운드마다 누가 누구를 잡았는지, 다음 목표가 누구인지 출력한다.<br>
다음은 출력 예시이다.
```
하윤 -서준(아웃)-> 지우
```

게임은 마지막 한 명이 남을 때까지 진행된다. 따라서 N명의 인원이 있는 경우 N-1번의 라운드가 진행된다.<br>
게임이 끝나면 마지막까지 살아남은 사람의 이름이 다음과 같이 출력된다.
```
우승자: XX
```

````python
# from random import random
# from random import shuffle

names = input("input: ").split()


shuffle(names)

for i in range(len(names)):
  print(names[i], end = " -> ")
print(names[0])

tail = {}
for i in range(len(names)):
  tail[names[i]] = names[(i+len(names)+1)%len(names)]

for i in range(len(names) - 1):
  tmp = names[int(random()*len(names))]
  names.remove(tail[tmp])
  print(f"{tmp} -{tail[tmp]}(아웃)-> {tail[tail[tmp]]}")
  tail[tmp] = tail[tail[tmp]]

print(f"우승자: {tmp}")
````

### Challenge: 꼬리 잡기 심화버전

꼬리잡기 시뮬레이션에서, 각 참가자가 목표를 잡은 횟수와 잡힌 순서를 저장할 수 있는 구조로 딕셔너리와 코드를 수정해 보자.
<br>
(힌트: 꼬리잡기 딕셔너리에는 목표, 잡은 횟수, 잡힌 순서 정보가 저장되어야 한다. 딕셔너리 안의 딕셔너리를 활용해 보자.)

````python
# from random import random
# from random import shuffle

names = input("참여 인원: ").split()
shuffle(names)

for i in range(len(names)):
  print(names[i], end = " -> ")
print(names[0])

tail = {}
for i in range(len(names)):
  tail[names[i]] = {'목표': names[(i+len(names)+1)%len(names)], '잡은횟수': 0, '잡힌 순서': 0}

for i in range(len(names) - 1):
  tmp = names[int(random()*len(names))]

  tail[tail[tmp]['목표']]['잡힌 순서'] = i+1
  tail[tmp]['잡은횟수'] += 1
  names.remove(tail[tmp]['목표'])

  print(f"{tmp} -{tail[tmp]['목표']}(아웃)-> {tail[tail[tmp]['목표']]['목표']}")
  tail[tmp]['목표'] = tail[tail[tmp]['목표']]['목표']

print(f"우승자: {tmp}")

print()
print(tail)
````
