## 📘 실습 이론 정리
### 1. 리스트란?
* 리스트(list)는 여러 개의 값을 하나의 변수에 순서대로 저장하는 자료형이다.
* 대괄호 []를 사용하여 생성한다.
* 리스트에는 문자열, 숫자, 실수 등 다양한 자료형을 함께 저장할 수 있다.
* 리스트는 **순서(index)**를 가지며, 중복된 값도 허용한다.
### 2. 리스트의 특징
* 순서가 있다
→ 각 원소는 인덱스(index)를 통해 접근 가능
* 수정 가능하다 (mutable)
→ 생성 후에도 값 변경 가능
* 중복 허용
* 여러 개의 값을 한 번에 관리할 수 있어 반복문과 함께 자주 사용된다
### 3. 리스트와 인덱스(Index)
* 리스트의 인덱스는 0부터 시작한다.
* 음수 인덱스를 사용하면 뒤에서부터 접근할 수 있다.
* 인덱스를 통해 특정 위치의 값을 읽거나 수정할 수 있다.

⚠️ 존재하지 않는 인덱스에 접근하면 오류(IndexError)가 발생한다.

### 4. 리스트의 길이
* len() 함수를 사용하면 리스트에 저장된 원소의 개수를 알 수 있다.
* 반복문에서 리스트를 다룰 때 자주 사용된다.
### 5. 리스트와 반복문
* 리스트는 반복문(for, while)과 함께 자주 사용된다.
* 리스트의 모든 원소를 하나씩 처리할 때 매우 유용하다.
* 반복문을 사용하면 리스트의 각 원소에 대해 동일한 작업을 수행할 수 있다.
### 6. 리스트에 값 추가 및 삭제
* 리스트는 실행 중에도 값의 추가와 삭제가 가능하다.
* 값 추가: 리스트의 끝에 새로운 값을 넣을 수 있다.
* 값 삭제: 특정 값이나 특정 위치의 값을 제거할 수 있다.

⚠️ 삭제 시 존재하지 않는 값을 제거하려 하면 오류가 발생할 수 있다.

###7. 리스트와 문자열
* 문자열은 문자의 리스트처럼 다룰 수 있다.
* 문자열을 공백 기준으로 나누면 리스트가 된다.
* 리스트의 각 원소는 문자열 처리와 조건문과 함께 자주 사용된다.

### 8. 리스트 정렬과 순서
* 리스트는 정렬하여 사용할 수 있다.
* 오름차순 또는 내림차순 정렬이 가능하다.
* 정렬 후에는 리스트의 원소 순서가 변경된다.
### 9. 리스트 사용 시 자주 하는 실수
* 인덱스를 1부터 시작한다고 착각하는 경우
* 리스트 길이를 넘는 인덱스에 접근하는 경우
* 문자열과 리스트를 혼동하는 경우
* 리스트를 복사하지 않고 그대로 참조해서 사용하는 경우
