### Problem 1: 로또 번호 생성기

😄 while문을 이용하여 로또번호 6개를 생성하는 프로그램을 작성하시오.
- 로또번호는 1~45사이 random.randint()를 사용
- 중복 번호를 허용함

---

Output Sample
```
로또번호 생성기 프로그램
1번째 번호는 40입니다.
2번째 번호는 11입니다.
3번째 번호는 11입니다.
4번째 번호는 28입니다.
5번째 번호는 31입니다.
6번째 번호는 5입니다.

산다. 산다. 집 산다!!!
```

````python
import random

print("로또번호 생성기 프로그램")

count = 0

while count < 6:  # 0,1,2,3,4,5일 때만 반복
    num = random.randint(1, 45)  # 1~45, 중복 허용
    count += 1
    print(f"{count}번째 번호는 {num}입니다.")

print()
print("산다. 산다. 집 산다!!!")
````

⚠️ 주의: while문의 반복 조건 count < 6에서 < 대신 <=이 쓰이면 어떻게 될지 살펴보도록 하자. 프로그램이 여섯 번째 번호를 출력하고 나서, count 값은 여전히 6이므로 while 문은 멈추지 않게 된다.
이제 count 값이 7로 올라가고 일곱 번째 번호가 출력되고 나서야 while문이 끝날 것이다.
이처럼 반복문은 반복 조건을 어떻게 설정하냐에 따라 결과가 크게 달라질 수 있어 주의가 필요하다.

````python
correct_pw = "예비대학"
count = 0

while count < 5:
    pw = input("암호를 입력하세요: ")
    count += 1

    if pw == correct_pw:
        print("You logged in!")
        break

if pw != correct_pw:
    print("5회 틀렸습니다. 로그인에 실패하셨습니다.")
````

📜참고: while True와 같이 반드시 성립하는 반복 조건 을 사용하는 경우 나중에 break 로 반복문을 끝내야 한다.

😄 아래 코드와 정확히 동일한 기능을 수행하는 프로그램을 while문 없이 작성하시오.

### Problem 6: 콜라츠 수
😄 독일의 수학자 로타르 콜라츠는 아래와 같은 추측을 제시하였다. 어떤 수가 있을 때, 이 수가 짝수면 2로 나누고 홀수면 3을 곱한 후 1을 더한다.
이 과정을 반복하다 보면 반드시 1 이 나오지 않을까?
콜라츠의 추측을 검증하기 위한 프로그램을 아래 예시와 같이 작성하시오.

---

<font color='blue'>`Test case-1`</font>

Input Sample
```
10
```
Output Sample
```
수를 입력하세요:
5
16
8
4
2
1
역시 콜라츠 대단하네요.
```

````python
n = int(input("수를 입력하세요: "))

while n != 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = n * 3 + 1
    print(n)

print("역시 콜라츠 대단하네요.")
````

### (과제) Problem 7: 목표 높이까지 최소의 블럭 개수로 탑 쌓기

😄 친구들과 블럭으로 탑을 쌓고 있어요.

목표 높이(cm)를 정확히 맞춰서 최소의 블럭 개수로 탑을 만들려고 합니다.
반복문을 사용하여 구하세요.

조건:
- 목표 높이는 100cm 이하의 정수로 입력됩니다.
- 사용할 수 있는 블럭의 각 크기(상대적으로 긴 방향의 길이)는 다음과 같습니다.
  - 15cm / 7cm / 3cm / 1cm
- 각 블럭은 무한개 있다고 가정합니다.
- 항상 큰 블럭부터 사용합니다.

---

<font color='blue'>`Test case-1`</font>

Input Sample
```
63
```
Output Sample
```
목표 높이:
최소 블럭 개수: 5
```
<font color='blue'>`Test case-2`</font>

Input Sample
```
78
```

Output Sample
```
목표 높이:
최소 블럭 개수: 6
```






````python
target = int(input("목표 높이: "))

count = 0

while target > 0:
  if target >= 15:
    target -= 15
    count += 1
  elif target >= 7:
    target -= 7
    count += 1
  elif target >= 3:
    target -= 3
    count += 1
  else:
    target -= 1
    count += 1

print("최소 블럭 개수:", count)
````

### (과제) Problem 8: 마법사의 주문 연습하기

🧙 마법사가 주문을 쓰면 마나가 소모됩니다.
마법사는 마나를 잘 관리하면서 콤보를 쌓아야 합니다.

아래처럼 마나(mana)는 10으로, 콤보(combo)는 0에서 시작합니다. 먼저 초기 상태를 출력해줍니다.
```
mana = 10
combo = 0
```

마법사는 한 턴에 주문을 한 번씩만 사용할 수 있습니다. 주문은 문자열로 입력됩니다.
- 🔥 불 주문("fire"): 마나 3 소모 / 콤보 1 증가
- ❄️ 얼음 주문("ice"): 마나 2 소모 / 콤보 1 증가
- 💧 마나 회복 주문("heal"): 마나 4 회복 / 콤보는 0으로 초기화
- ⚠️ 단, 주문을 잘못 입력해도 콤보가 초기화되고 마나가 조금 소모(-2)됩니다.


만약, 연습 중지("stop" 입력)를 하거나, 마나가 부족(*주문을 반영했을 때 마나가 0 미만이 되는 것인지 사전 체크 필요!!*)하면 주문이 실행되기 전에 즉시 주문 연습이 종료됩니다.

주문 연습이 종료되면, 종료 시점의 콤보 수를 출력합니다.

---

<font color='blue'>`Test case-1`</font>

Input Sample
```
ice
heal
fire
stop
```
Output Sample
```
현재 마나: 10 / 콤보: 0
주문: ice
얼음 주문!
현재 마나: 8 / 콤보: 1
주문: heal
회복 주문!
현재 마나: 12 / 콤보: 0
주문: fire
불 주문!
현재 마나: 9 / 콤보: 1
주문: stop
연습 종료!
최종 콤보: 1
```

<font color='blue'>`Test case-2`</font>

Input Sample
```
fire
ice
fire
achu
ice
```
Output Sample
```
현재 마나: 10 / 콤보: 0
주문:
불 주문!
현재 마나: 7 / 콤보: 1
주문:
얼음 주문!
현재 마나: 5 / 콤보: 2
주문:
불 주문!
현재 마나: 2 / 콤보: 3
주문:
잘못된 주문!
현재 마나: 0 / 콤보: 0
주문:
마나 부족...
연습 종료!
최종 콤보: 0
```

````python
mana = 10
combo = 0
print("현재 마나:", mana, "/ 콤보:", combo)

while True:
    spell = input("주문: ")

    if spell == "stop":
      break

    # 불 주문
    if spell == "fire":
      if mana < 3: # 주문을 반영했을 때 마나가 0 미만이 되는 것인지 사전 체크 필요!!
        print('마나 부족...')
        break
      mana -= 3
      combo += 1
      print("불 주문!")

    # 얼음 주문
    elif spell == "ice":
      if mana < 2: # 주문을 반영했을 때 마나가 0 미만이 되는 것인지 사전 체크 필요!!
        print('마나 부족...')
        break
      mana -= 2
      combo += 1
      print("얼음 주문!")

    # 회복 주문
    elif spell == "heal":
      mana += 4
      combo = 0
      print("회복 주문!")

    # 잘못된 주문
    else:
      if mana < 2:
        print('마나 부족...') # 주문을 반영했을 때 마나가 0 미만이 되는 것인지 사전 체크 필요!!
        break
      mana -= 2
      combo = 0
      print("잘못된 주문!")

    print("현재 마나:", mana, "/ 콤보:", combo)

print("연습 종료!")
print("최종 콤보:", combo)
````
